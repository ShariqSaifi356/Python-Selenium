```python
# Selection Sort

arr = [3,4,50,2,3,1]

n = len(arr) - 1

for i in range(n):
    temp = i
    for j in range(i+1, n + 1):
        if arr[j] < arr[temp]:
            temp = j
    arr[i], arr[temp] = arr[temp], arr[i]
    
print(arr)
```

```python
# Decorators

def decorator_function(main_function):
    def wrapper(a, b):
        print("Before Operation")
        result = main_function(a, b)
        print("After Operation")
        return result
    return wrapper

@decorator_function
def add(a, b):
    return a + b

print(add(20, 10))

@decorator_function
def sub(a, b):
    return a - b

print(sub(20, 10))
```

```python
# Prime Number

def is_prime(num : int) -> bool:
    
    if num <= 1:
        return False
    
    if num == 2:
        return True
    
    if num % 2 == 0:
        return False
    
    temp = 3
    
    while temp <= num:
        if num % temp == 0:
            return False
        temp = temp + 2
        
    return True

```

```python
def fibo(num):
    
    if num < 0:
        return -1
    
    elif num == 0:
        return 0
    
    elif num == 1 or num == 2:
        return 1
    
    else:
        return fibo(num - 1) + fibo(num - 2)
    
n = 20
print(fibo(n))

for i in range(n):
    print(fibo(i), end=" ")
```

```python
# Fibonacci Series.

def fibonacci(num):
    if num < 0:
        return -1
    elif num == 0:
        return 0
    elif num == 1 or num == 2:
        return 1
    else:
        return fibonacci(num -1) + fibonacci(num - 2)
    
num = 10
for i in range(num+1):
    print(fibonacci(i), end = " ")
```

```python
# Factorial

def fact(number : int) -> int:
    if number <= 0:
        return -1
    factorial = 1
    for i in range(2, number+1):
        factorial = factorial * i
    return factorial

num = 5
print(fact(num))
```

```python
# Armstrong Number.
import math as m

def armstrong_number(number : int)->bool:
    if number < 0:
        return False
    
    count = len(str(number))
    temp_number = number
    arm_number = 0
    
    while temp_number != 0:
        digit = temp_number % 10
        arm_number = arm_number + int(m.pow(digit, count))
        temp_number //= 10
    return arm_number == number

num = 153
print(armstrong_number(num))

```

```python
# Automorphic Number
import math
def automorphic_number(number : int) -> bool:
    if number < 0:
        return False
    
    count = len(str(number))
    square = number**2
    last_digits = square % int(math.pow(10, count))
    
    return last_digits == number

num = 25
print(automorphic_number(num))
        
```

```python
# Find Second Largest Number Without Sorting

num = [1, 4, 50, 6, 7, 9]

largest = num[0]
second_largest = num[0]

for n in num:
    if n > largest:
        second_largest = largest
        largest = n
    elif n > second_largest and n != largest:
        second_largest = n

print("Second Largest:", second_largest)
```

1. map()

Purpose: Apply the same operation to every element.

Example: Square all numbers
lst = [1, 2, 3, 4]

result = map(lambda x: x**2, lst)

print(list(result))

Output:

[1, 4, 9, 16]
Simple Meaning
Take every item
↓
Do something to it
↓
Return the modified items
2. filter()

Purpose: Keep only elements that satisfy a condition.

Example: Keep even numbers
lst = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, lst)

print(list(result))

Output:

[2, 4, 6]
Simple Meaning
Check each item
↓
Condition True → Keep it
Condition False → Remove it
Difference Between map() and filter()
lst = [1, 2, 3, 4]
map()
map(lambda x: x*10, lst)

Output:

[10, 20, 30, 40]

Changes values.

filter()
filter(lambda x: x % 2 == 0, lst)

Output:

[2, 4]

Removes values.

3. zip()

Purpose: Combine multiple iterables element by element.

Example
names = ["Dhoni", "Rohit", "Virat"]
scores = [95, 88, 99]

result = zip(names, scores)

print(list(result))

Output:

[('Dhoni', 95), ('Rohit', 88), ('Virat', 99)]
Simple Meaning
First with First
Second with Second
Third with Third

Like a zipper joining two sides together.

Real Example
names = ["A", "B", "C"]
ages = [20, 21, 22]

for name, age in zip(names, ages):
    print(name, age)

Output:

A 20
B 21
C 22
4. enumerate()

Purpose: Get both index and value while looping.

Without enumerate()
lst = ["Python", "Java", "C++"]

for i in range(len(lst)):
    print(i, lst[i])

Output:

0 Python
1 Java
2 C++
With enumerate()
lst = ["Python", "Java", "C++"]

for index, value in enumerate(lst):
    print(index, value)

Output:

0 Python
1 Java
2 C++
Simple Meaning
Give me:
1. Position (index)
2. Value

at the same time.

map() → Applies a function to every element of an iterable.

filter() → Returns only the elements that satisfy a condition.

zip() → Combines multiple iterables element-wise into tuples.

enumerate() → Returns both index and value while iterating.


Iterable vs. Iterator (The Critical Difference)

People often confuse these two terms, but they are different steps in the process:

Iterable: Anything you can loop over (like a List, Tuple, or String). It has an __iter__() method that manufactures an iterator, but it cannot fetch items on its own using next().

Iterator: The actual engine doing the traversing. It has a __next__() method to fetch items one by one.
