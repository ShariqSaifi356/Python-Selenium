```python
# 1. Find frequency of each element

arr = [2,3,2,4,2,2,5,6,6,7]
for i in set(arr):
    print(f"Count of {i} is : {arr.count(i)}")
```

```python
# 2. Print duplicate elements

arr = [2,3,2,4,2,2,5,6,6,7]
for i in set(arr):
    if arr.count(i) > 1:
        print(i)
```

```python
# 3. Print array in reverse order

arr = [10,20,30,40]
print(arr[::-1])
```

```python
arr = [10, 20, 30, 40, 50]
rev = []

for i in range(len(arr)-1, -1, -1):
    rev.append(arr[i])
    
print(rev)
```

```python
# 4. Find the largest element
arr = [3,4,50,2,3,1]
print(max(arr))
```

```python
arr = [3,4,50,2,3,1]
max_ele = arr[0]

for i in arr:
    if max_ele < i:
        max_ele = i
        
print(max_ele)
```

```python
# 5. Find the smallest element

arr = [3,4,50,2,3,1]
print(min(arr))
```

```python
arr = [3,4,50,2,3,1]
min_ele = arr[0]

for i in arr:
    if min_ele > i:
        min_ele = i
        
print(min_ele)
```

```python
# Print sum of all items

arr = [10, 20, 30, 40]
print(sum(arr))
```

```python
arr = [10, 20, 30, 40]
addition = 0

for i in arr:
    addition += i
    
print(addition)
```

```python
# Sort elements in ascending order
arr = [3,4,50,2,3,1]
arr.sort()
print(arr)
```

```python
# Selection Sort

arr = [3,4,50,2,3,1]

n = len(arr) - 1

for i in range(n):
    temp = i
    for j in range(i+1, n + 1):
        if arr[j] < arr[temp]:
            temp = j
    arr[i], arr[temp] = arr[temp], arr[i]
    
print(arr)
```

```python
# 8. Find the 2nd largest number
arr = [3,4,50,2,3,1]
arr = sorted(arr)
print(arr[-2])
```

```python
# 9. Find the 2nd samllest number
arr = [3,4,50,2,3,1]
arr = sorted(arr)
print(arr[1])
```

```python
# 10. Separate odd and even numbers

arr = [3,4,50,2,3,1]
even = []
odd = []

for i in arr:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
        
print(even)
print(odd)
```

```python
# 11. Remove all white spaces 

text = "Hello My Dear Friends"
print(text.replace(" ", ""))
```
