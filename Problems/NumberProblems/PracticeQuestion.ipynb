```python
# 1. Find frequency of each element

arr = [2,3,2,4,2,2,5,6,6,7]
for i in set(arr):
    print(f"Count of {i} is : {arr.count(i)}")
```

```python
# 2. Print duplicate elements

arr = [2,3,2,4,2,2,5,6,6,7]
for i in set(arr):
    if arr.count(i) > 1:
        print(i)
```

```python
# 3. Print array in reverse order

arr = [10,20,30,40]
print(arr[::-1])
```

```python
# Sort elements in ascending order
arr = [3,4,50,2,3,1]
arr.sort()
print(arr)
```

```python
# Selection Sort

arr = [3,4,50,2,3,1]

n = len(arr) - 1

for i in range(n):
    temp = i
    for j in range(i+1, n + 1):
        if arr[j] < arr[temp]:
            temp = j
    arr[i], arr[temp] = arr[temp], arr[i]
    
print(arr)
```

```python
# 10. Separate odd and even numbers

arr = [3,4,50,2,3,1]
even = []
odd = []

for i in arr:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
        
print(even)
print(odd)
```

```python
# Inheritance

class A:
    def add1(self,a,b):
        return a+b
    
class B(A):
    def minus(self,a,b,c):
        return self.add1(a,b) - c
    
obj1 = A()
obj2 = B()

print(obj1.add1(10,20))
print(obj2.minus(10, 20, 5))
        
```

```python
# Polymorphism

class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")
    
    
ob1 = Dog()
ob2 = Cat()

ob1.sound()
ob2.sound()
```

```python
# Encapsulation

class Bank:
    def __init__(self):
        self.__balance = 1000
        
    def show_balance(self):
        return self.__balance
        
ob3 = Bank()
print(ob3.show_balance())
```

```python
# A generator is a special function that uses the yield keyword to produce values one at a time instead of creating and storing all values in memory at once.


def numbers():
    for i in range(1, 6):
        yield i
        
num = numbers()

print(next(num))
print(next(num))
```

```python
# Diff b/w append() and extend()

lst1 = [1,2,3,4,5]
lst1.append([6,7])
print(lst1)
```

```python
# Diff b/w append() and extend()

lst1 = [1,2,3,4,5]
lst1.extend([6,7])
print(lst1)
```

```python
# list comprehension
# Syntax : [expression for item in iterable if condition]
lst1 = [1,2,3,4,5]
lst2 = [x**2 for x in lst1]
print(lst2)
```
