```python
import requests

Base_URL = "https://dummyjson.com"
```

 Verify successful GET returns correct schema

```python
response  = requests.get(f"{Base_URL}/users/1", timeout=5)
assert response.status_code == 200
body = response.json()

for i in ["id", "firstName", "lastName", "email", "age"]:
    assert i in body
```

2. Negative test — invalid/non-existent resource ID

```python
response  = requests.get(f"{Base_URL}/users/991", timeout=5)
assert response.status_code == 404
# print(response.json())
# Output : {'message': "User with id '991' not found"}
assert response.json()["message"] == "User with id '991' not found"
```

```python
response  = requests.get(f"{Base_URL}/users/1", timeout=5)
body = response.json()
print("Status: ", response.status_code)
print("Url : ", response.url)
print(len(body))

for i in body:
    print(f"{i} : {body[i]}")
    
    
print("-----------------------------------")
print(body["address"]["city"])
```

Pagination correctness

```python
response = requests.get(f"{Base_URL}/users", params={"limit":5, "skip": 10}, timeout=5)
body = response.json()
# print(body["skip"])
assert body["limit"] == 5
assert body["skip"] == 10
```
